```python
import pandas as pd
import numpy as np
```

```python
# Load the raw sustainability data
sust_raw = pd.read_csv("UNSust.csv")
print(f"Original shape: {sust_raw.shape}")
print("\nFirst few rows:")
print(sust_raw.head())
```

```python
print("Column names:")
print(sust_raw.columns.tolist())
print("\nData types:")
print(sust_raw.dtypes)
print("\nMissing values:")
print(sust_raw.isnull().sum())
```

```python
# remove Value Footnotes columns
sust_clean = sust_raw.drop(
    columns=["Value Footnotes", "Value Footnotes.1"], errors="ignore"
)
print(f"Shape after dropping footnotes: {sust_clean.shape}")
print("\nRemaining columns:")
print(sust_clean.columns.tolist())
```

4. Remove Regional Aggregates and Non-Country Entities

Following the same approach as the Life Expectancy dataset, we'll remove regional groupings and keep only individual countries.

```python
# unique country/area names we have
print(f"Number of unique countries/areas: {sust_clean['Country or Area'].nunique()}")
print("\nSample of country names:")
print(sorted(sust_clean["Country or Area"].unique())[:100])
```

```python
print(f"Total unique countries/areas: {sust_raw['Country or Area'].nunique()}")
print("\nAll country/area names:")
for name in sorted(sust_raw["Country or Area"].unique()):
    print(f"  {name}")
```

```python
# exact list of non-country entities to remove
non_countries = [
    "Africa Eastern and Southern",
    "Africa Western and Central",
    "Arab World",
    "Caribbean small states",
    "Early-demographic dividend",
    "East Asia & Pacific",
    "East Asia & Pacific (IDA & IBRD)",
    "East Asia & Pacific (excluding high income)",
    "Europe & Central Asia",
    "Europe & Central Asia (IDA & IBRD)",
    "Europe & Central Asia (excluding high income)",
    "Fragile and conflict affected situations",
    "Heavily indebted poor countries (HIPC)",
    "High income",
    "IBRD only",
    "IDA & IBRD total",
    "IDA blend",
    "IDA only",
    "IDA total",
    "Late-demographic dividend",
    "Latin America & Caribbean",
    "Latin America & Caribbean (IDA & IBRD)",
    "Latin America & Caribbean (excluding high income)",
    "Least developed countries: UN classification",
    "Low & middle income",
    "Low income",
    "Lower middle income",
    "Middle East & North Africa",
    "Middle East & North Africa (IDA & IBRD)",
    "Middle East & North Africa (excluding high income)",
    "Middle income",
    "Other small states",
    "Pacific island small states",
    "Post-demographic dividend",
    "Pre-demographic dividend",
    "Small states",
    "South Asia",
    "South Asia (IDA & IBRD)",
    "Sub-Saharan Africa",
    "Sub-Saharan Africa (IDA & IBRD)",
    "Sub-Saharan Africa (excluding high income)",
    "Upper middle income",
    "World",
]

# filter out non-countries
print(f"Rows before filtering: {len(sust_clean)}")
sust_clean = sust_clean[~sust_clean["Country or Area"].isin(non_countries)]
print(f"Rows after filtering: {len(sust_clean)}")

# look at the remaining countries
print(
    f"\nNumber of unique countries after filtering: {sust_clean['Country or Area'].nunique()}"
)
print("\nRemaining countries:")
for name in sorted(sust_clean["Country or Area"].unique()):
    print(f"  {name}")
```

```python
# common encoding issues
replacements = {
    "CÃ´te d'Ivoire": "Côte d'Ivoire",
    "SÃ£o TomÃ© and Principe": "São Tomé and Principe",
    "CuraÃ§ao": "Curaçao",
    "RÃ©union": "Réunion",
    "TÃ¼rkiye": "Turkey",
}

sust_clean["Country or Area"] = sust_clean["Country or Area"].replace(replacements)
```

```python
# range of values
print(f"\nValue statistics:")
print(sust_clean["Value"].describe())
print(f"\nYear range: {sust_clean['Year'].min()} to {sust_clean['Year'].max()}")
```

```python
# Sort by Country and Year
sust_clean = sust_clean.sort_values(["Country or Area", "Year"]).reset_index(drop=True)

print("\nFinal cleaned dataset preview:")
print(sust_clean.head(10))
print(f"\nFinal shape: {sust_clean.shape}")
```

```python
# summary stats
print(f"\nNumber of countries: {sust_clean['Country or Area'].nunique()}")
print(f"Number of years covered: {sust_clean['Year'].nunique()}")
print(f"Year range: {sust_clean['Year'].min()} - {sust_clean['Year'].max()}")
print(f"Total observations: {len(sust_clean)}")
print(
    f"\nSustainability value range: {sust_clean['Value'].min():.2f} - {sust_clean['Value'].max():.2f}"
)
print(f"Mean sustainability: {sust_clean['Value'].mean():.2f}")
print(f"Median sustainability: {sust_clean['Value'].median():.2f}")
```

```python
output_file = "UNSust_clean.csv"
sust_clean.to_csv(output_file, index=False)
print(f"\nCleaned dataset saved to: {output_file}")
```
