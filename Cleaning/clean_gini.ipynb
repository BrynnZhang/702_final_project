```python
import pandas as pd

gini = pd.read_csv("UNGini.csv")
```

```python
print(gini)
```

```python
gini = gini.drop("Value Footnotes", axis=1)
gini = gini.drop("Value Footnotes.1", axis=1)
print(gini)
```

```python
gini = gini[~gini["Country or Area"].str.isnumeric()]
gini = gini[:-1]
print(gini)
```

```python
gini = gini.dropna()
print(gini)
```

```python
print(gini["Country or Area"].unique())
```

```python
gini["Country or Area"] = gini["Country or Area"].replace(
    "CÃ´te d'Ivoire", "Côte d'Ivoire"
)
gini["Country or Area"] = gini["Country or Area"].replace("Korea", "Republic of Korea")
gini["Country or Area"] = gini["Country or Area"].replace(
    "Congo", "Democratic Republic of the Congo"
)
gini["Country or Area"] = gini["Country or Area"].replace("The Gambia", "Gambia")
gini["Country or Area"] = gini["Country or Area"].replace(
    "Kyrgyz Republic", "Kyrgyzstan"
)
gini["Country or Area"] = gini["Country or Area"].replace(
    "Lao PDR", "Lao People's Democratic Republic"
)
gini["Country or Area"] = gini["Country or Area"].replace(
    "Moldova", "Republic of Moldova"
)
gini["Country or Area"] = gini["Country or Area"].replace(
    "Russia", "Russian Federation"
)
gini["Country or Area"] = gini["Country or Area"].replace("St. Lucia", "Saint Lucia")
gini["Country or Area"] = gini["Country or Area"].replace(
    "SÃ£o TomÃ© and Principe", "Sao Tome and Principe"
)
gini["Country or Area"] = gini["Country or Area"].replace("Slovak Republic", "Slovakia")
gini["Country or Area"] = gini["Country or Area"].replace("TÃ¼rkiye", "Türkiye")
gini["Country or Area"] = gini["Country or Area"].replace(
    "United Kingdom", "United Kingdom of Great Britain and Northern Ireland"
)
gini["Country or Area"] = gini["Country or Area"].replace(
    "Tanzania", "United Republic of Tanzania"
)
gini["Country or Area"] = gini["Country or Area"].replace(
    "United States", "United States of America"
)
```
