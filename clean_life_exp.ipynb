```python
import pandas as pd

life_exp = pd.read_csv("UNLifeExp.csv")
```

```python
print(life_exp)
```

```python
life_exp = life_exp.drop("Variant", axis=1)
print(life_exp)
```

```python
life_exp = life_exp.dropna()
print(life_exp)
```

```python
print(life_exp["Country or Area"].unique())
```

```python
life_exp = life_exp[life_exp["Country or Area"] != "Micronesia"]

life_exp["Country or Area"] = life_exp["Country or Area"].replace("CuraÃ§ao", "Curaçao")
life_exp["Country or Area"] = life_exp["Country or Area"].replace(
    "CÃ´te d'Ivoire", "Côte d'Ivoire"
)
life_exp["Country or Area"] = life_exp["Country or Area"].replace(
    "Bolivia (Plurinational State of)", "Bolivia"
)
life_exp["Country or Area"] = life_exp["Country or Area"].replace(
    "Venezuela (Bolivarian Republic of)", "Venezuela"
)
life_exp["Country or Area"] = life_exp["Country or Area"].replace(
    "Iran (Islamic Republic of)", "Iran"
)
life_exp["Country or Area"] = life_exp["Country or Area"].replace(
    "Micronesia (Fed. States of)", "Micronesia"
)
life_exp["Country or Area"] = life_exp["Country or Area"].replace("TÃ¼rkiye", "Türkiye")
life_exp["Country or Area"] = life_exp["Country or Area"].replace(
    "Dem. People's Republic of Korea", "Democratic People's Republic of Korea"
)

life_exp["Country or Area"] = life_exp["Country or Area"].replace(
    "United Kingdom", "United Kingdom of Great Britain and Northern Ireland"
)
```

```python
countries = pd.read_excel("un_countries.xlsx")
countries = countries["English short"]
countries = countries[:-7]
countries = countries.str.replace(r"\(.*?\)", "", regex=True).str.strip()
```
